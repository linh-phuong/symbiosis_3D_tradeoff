```python
import matplotlib.pylab as plt
from tools import turn_raw_dat_to_sorted_arrays
import pickle
import numpy as np
import pandas as pd
```

```python
with open("data_evol_traject/ds1_scenario1a_exp.pkl", "rb") as f:
    ds_1 = pickle.load(f)
dss_1 = turn_raw_dat_to_sorted_arrays(ds_1)
```

```python
popmin_thres = 5
fig, axes = plt.subplots(1, 3, figsize = (15, 5))
cc = np.where(np.sum(dss_1["xF_rho"] > popmin_thres, axis = 0) > 0)[0]
ZZ = dss_1["xF_rho"][:, cc]
extent = [dss_1["rho_sort"][cc][0], dss_1["rho_sort"][cc][-1], ds_1["t"][0], ds_1["t"][-1]]
vmax = np.sqrt(ZZ.max() * ZZ[ZZ > 0].min())
axes[0].imshow(ZZ, cmap = "Greys", aspect = "auto", extent = extent, origin = "lower",
               vmax = vmax)
axes[0].set_title(r"Independent reproduction $\rho$")
axes[0].set_xlim(0, 10)
cc = np.where(np.sum(dss_1["xF_nu"] > popmin_thres, axis = 0) > 0)[0]
ZZ = dss_1["xF_nu"][:, cc]
vmax = np.sqrt(ZZ.max() * ZZ[ZZ > 0].min())
extent = [dss_1["nu_sort"][cc][0], dss_1["nu_sort"][cc][-1], ds_1["t"][0], ds_1["t"][-1]]
axes[1].imshow(ZZ, cmap = "Greys", aspect = "auto", extent = extent, origin = "lower",
               vmax = vmax)
axes[1].set_title(r"Symbiont effect $\nu$")
axes[1].set_xlim(-1.5, 3)
cc = np.where(np.sum(dss_1["xF_tau"] > popmin_thres, axis = 0) > 0)[0]
ZZ = dss_1["xF_tau"][:, cc]
vmax = np.sqrt(ZZ.max() * ZZ[ZZ > 0].min())
extent = [dss_1["tau_sort"][cc][0], dss_1["tau_sort"][-1], ds_1["t"][0], ds_1["t"][-1]]
axes[2].imshow(ZZ, cmap = "Greys", aspect = "auto", extent = extent, origin = "lower",
               vmax = vmax)
axes[2].set_title(r"Bound reproduction $\tau$")
axes[2].set_xlim(0, 10)
```

```python
ds1_exp = pd.DataFrame(dict(nu = ds_1["nu"][-1], rho = ds_1["rho"][-1]))
ds1_exp = ds1_exp.iloc[list(range(0, ds1_exp.shape[0], 1)), :]
ds1_exp.to_csv("data_evol_traject/ds1_scenario1a_exp.csv", index = False, header=False)
```
